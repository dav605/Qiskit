# 1. Inspecting a Quantum State Using `Statevector`

In this section, we extract and inspect key properties of a 1-qubit quantum state:
* **Amplitudes:** The complex values representing each state.
* **Probabilities:** The square of the absolute values of the amplitudes.
* **State Vector:** The complete state representation.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
qc=QuantumCircuit(1)
qc.x(0)
state=Statevector.from_instruction(qc)
print("State: ",state)
amplitude=state.data
print("Amplitude: ",amplitude)
probabilities=state.probabilities()
print("Probabilities: ",probabilities)
print("States: ",state)
```

# 2. Inspecting the Full Unitary Matrix Using `Operator`

Here, we represent a quantum circuit as a full unitary matrix mapping input states to output states.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator
qc=QuantumCircuit(1)
qc.h(0)
print(qc.draw())
unitary=Operator(qc)
print(unitary)
```

# 3. Verifying Qiskit Output Against NumPy

To ensure our understanding of the under-the-hood mathematics, we manually construct the logic using raw NumPy matrices and compare the resulting state and unitary operators.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector,Operator
import numpy as np
qc=QuantumCircuit(1)
qc.x(0)
qc.z(0)
print(qc.draw())
state=Statevector.from_instruction(qc)
print("State: ",state)
unitary=Operator(qc)
print(unitary)
#numpy
X=np.array([[0,1],
            [1,0]],dtype=complex)
Z=np.array([[1,0],
            [0,-1]],dtype=complex)
initial_state=np.array([1,0],dtype=complex)
numpy_unitary=Z@X
numpy_state=numpy_unitary @ initial_state
print(numpy_state)
print("States match",np.allclose(state,numpy_state))
print("operators match",np.allclose(unitary,numpy_unitary))
```

# 4. Verifying Operator Unitarity

We verify that the obtained operator $U$ is indeed unitary by confirming that $U^{\dagger} U = I$ holds true.

```python
! pip install qiskit
from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator
qc=QuantumCircuit(2)
qc.h(0)
qc.z(0)
U=Operator(qc).data
print(U)
U_dagger=U.conj().T
print(np.allclose(U_dagger @ U,np.eye(4)))
```

# 5. Inspecting a Two-Qubit Circuit

Now, we extend our verification to a 2-qubit system containing an entangling gate (CNOT) and check its unitarity.

```python
! pip install numpy
from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator
import numpy as np
qc=QuantumCircuit(2)
qc.h(0)
qc.cx(0,1)
U=Operator(qc).data
U_dagger=U.conj().T
print("Is the matrix Unitary:")
print(np.allclose(U_dagger@U,np.eye(4)))
```
