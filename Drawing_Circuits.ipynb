### 4. Circuit Depth

```python
from qiskit import QuantumCircuit
from IPython.display import display
qc=QuantumCircuit(2)
qc.x(0)
qc.x(1)
qc.cx(0,1)
display(qc.draw(output="mpl"))
qc.depth()
```

### 5. Count the Number of Operations (`count_ops`)

```python
from qiskit import QuantumCircuit
qc=QuantumCircuit(2)
qc.x(0)
qc.x(1)
qc.cx(0,1)
qc.count_ops()
```
