### Workflow for Parameterized Circuits

1. Define a symbolic parameter.
2. Initialize a quantum circuit.
3. Apply a parameterized rotation gate to the circuit.
4. Display the resulting circuit structure.
5. Inspect the parameters contained within the circuit.

```python
from qiskit import QuantumCircuit
from qiskit.circuit import Parameter
theta=Parameter("theta")
qc=QuantumCircuit(1)
qc.ry(theta,0)
print(qc)
print(qc.parameters)
```

### Using Parameter Vectors

When scaling to larger quantum circuits, managing individual parameters becomes tedious. We can use **ParameterVector** to register an array of parameters with a single definition.

```python
from qiskit import QuantumCircuit
from qiskit.circuit import ParameterVector
theta=ParameterVector("theta",3)
qc=QuantumCircuit(3)
qc.ry(theta[0],0)
qc.ry(theta[1],1)
qc.ry(theta[2],2)
print(qc)
print(qc.parameters)
```

### Assigning Parameter Values

To run or evaluate a parameterized circuit, we bind numerical values to the symbolic parameters using **assign_parameters**.

```python
from qiskit import QuantumCircuit
from qiskit.circuit import Parameter
theta=Parameter("theta")
qc=QuantumCircuit(1)
qc.ry(theta,0)
bound_qc=qc.assign_parameters(
    {"theta":1.57}
)
print(bound_qc)
```

### Binding Values to a Parameter Vector

Similarly, we can map numerical values to elements of **ParameterVector** using a dictionary.

```python
from qiskit import QuantumCircuit
from qiskit.circuit import ParameterVector
theta=ParameterVector("theta",3)
qc=QuantumCircuit(3)
qc.ry(theta[0],0)
qc.ry(theta[1],1)
qc.ry(theta[2],2)
bound_qc=qc.assign_parameters({
    theta[0]:0.5,
    theta[1]:1.0,
    theta[2]:1.5
})
print(qc.parameters)
print(bound_qc)
```

### Exercise: Three-Qubit Parameterized Circuit

**Objective:** Create a three-qubit circuit with parameterized rotations on different axes (X, Y, and Z) and assign distinct angles to each.

```python
from qiskit import QuantumCircuit
from qiskit.circuit import ParameterVector
theta=ParameterVector("theta",3)
qc=QuantumCircuit(3)
qc.rx(theta[0],0)
qc.ry(theta[1],1)
qc.rz(theta[2],2)
bound_qc=qc.assign_parameters(
    {
        theta[0]:0.5,
        theta[1]:1.0,
        theta[2]:1.5
    }
)
print(bound_qc)
```
