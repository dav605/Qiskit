# Qiskit Bit Ordering: Little-Endian

In this notebook, we explore how Qiskit represents quantum states and bit ordering.

### 1. The Concept of Bit Ordering

Typically, we might write a 3-qubit state from left to right as $|q_0 q_1 q_2\rangle$. However, Qiskit utilizes **little-endian** bit ordering.

This means that the lease-significant qubit ($q_0$) is stored on the **right-most** side of the state string:
$$\lvert q_2 q_1 q_0 \rangle$$

For example, if we apply an $X$ (NOT) gate to $q_0$ and $q_2$ (leaving $q_1$ as $|0\rangle$):
* $q_0 = 1$
* $q_1 = 0$
* $q_2 = 1$

The resulting state in memory is represented as **`101`**.

### 2. Verification using Statevector (Probability Vector)

```python
! pip install qiskit
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
qc=QuantumCircuit(3)
qc.x(0)
qc.x(2)
print(qc.draw())
state=Statevector.from_instruction(qc)
print("probablities",state.probabilities_dict())
```

### 3. Verification using Qiskit Aer Simulator

```python
! pip install qiskit-aer
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
qc=QuantumCircuit(3,3)
qc.x(0)
qc.x(2)
qc.measure([0,1,2],[0,1,2])
aer=AerSimulator()
results=aer.run(qc,shots=50).result()
count=results.get_counts()
print(count)
```
