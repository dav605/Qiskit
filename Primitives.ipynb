### Qiskit Primitives

There are two main types of primitives in Qiskit:
1. Sampler
2. Estimator

#### 1. Sampler

The Sampler primitive is used to collect bitstrings and reconstruct probability distributions.

We will use a Transverse Field Ising Model to demonstrate Qiskit's primitives in action.

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0,1)
qc.measure_all()
sampler=StatevectorSampler()
job=sampler.run([qc])
result=job.result()
#access the first PUB
first_pub=result[0]
# access the data
data = first_pub.data
# access measurement
meas = data.meas
#mea to count
count=meas.get_counts()
print(count)
```

#### 2. Estimator

The Estimator primitive computes the expectation values of observables for given quantum states.

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorEstimator
from qiskit.quantum_info import SparsePauliOp
qc=QuantumCircuit(1)
qc.x(0)
observable=SparsePauliOp("Z")
estimator=StatevectorEstimator()
job=estimator.run([(qc,observable)])
result= job.result()
first_pub=result[0]
evs=first_pub.data.evs
print(evs)
```

### PUB (Primitive Unified Block) Format

Inputs to Qiskit Primitives must be formatted as PUBs (Primitive Unified Blocks).

#### Sampler PUB Format:
(circuit, parameter_values, shots)

```python
job=sampler.run([(qc,None,50)])
```

#### Estimator PUB Format:
(circuit, observable, parameter_values, precision)

```python
job=estimator([(qc,observable,None,0.01)])
```

### Exercise: Sampler

Create a 2-qubit circuit that:
1. Applies an H gate to qubit 0.
2. Applies a CX gate from qubit 0 to qubit 1.
3. Measures both qubits.
4. Uses StatevectorSampler.
5. Executes the circuit for 1,000 shots.
6. Accesses result[0].
7. Prints the counts.

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
qc=QuantumCircuit(2)
qc.h(0)
qc.cx(0,1)
qc.measure_all()
sampler=StatevectorSampler()
job=sampler.run([qc],shots=1000)
result=job.result()
first_pub=result[0]
data=first_pub.data
meas=data.meas
counts=meas.get_counts()
print(counts)
```
