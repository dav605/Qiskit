2.Create a 2 Qubit gate


```python
from qiskit import QuantumCircuit
qc = QuantumCircuit(2)
print(qc)
```

3.Add Hadamard gate at 0 Qubit

```python
from qiskit import QuantumCircuit
qc = QuantumCircuit(2)
qc.h(0)
print(qc)
```

4.Add a X gate at 1 Qubit (It acts like classical NOT)

```python
from qiskit import QuantumCircuit
qc = QuantumCircuit(2)
qc.h(0)
qc.x(1)
print(qc)
```

5. Add CNOT instead of X 0 acts as control Qubit and 1 as target Qubit

```python
from qiskit import QuantumCircuit
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0,1)
print(qc)
```

6. Create a bell state (Maximally entangled pair) 

```python
from qiskit import QuantumCircuit
qc=QuantumCircuit(2)
qc.h(0)
qc.cx(0,1)
print(qc)
```

7. Measure the qubits (Measure qubit 0 and store the result in classical bit 0.)



```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(2, 2)

qc.h(0)
qc.cx(0, 1)

qc.measure(0, 0)
qc.measure(1, 1)

print(qc)
```

8. Instead of measure use measure_all


```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(2, 2)

qc.h(0)
qc.cx(0, 1)
qc.measure_all()
print(qc)
```

9. Rotation gates

```python
from qiskit import QuantumCircuit
import numpy as np
qc = QuantumCircuit(1)
qc.rx(np.pi/2, 0)
qc.ry(np.pi/2, 0)
qc.rz(np.pi/2, 0)
print(qc)
```

10. Circuit with 3 gates

```python
from qiskit import QuantumCircuit
qc= QuantumCircuit(3)
qc.h(0)
qc.cx(0,2)
qc.cx(1,2)
print(qc)
```

11. Circuit Information 

```python
print("Number of qubits",qc.num_qubits)
print("Depth of the circuit",qc.depth())
print("Size of the circuit",qc.size())
print("No of operation",qc.count_ops())
print("Classical bits:", qc.num_clbits)
```

12. State vectors(No of likelyhood probablities)

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
state=Statevector.from_instruction(qc)
print(state) #1/root(2)=0.707 
#4 states
```
