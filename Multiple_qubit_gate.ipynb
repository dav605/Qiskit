1.Create a 2-qubit circuit. Set the first qubit to 1 and apply a CX gate with the first qubit as control and the second qubit as target. Find the final state.

```python
!pip install qiskit
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
qc=QuantumCircuit(2)
qc.x(0)
qc.cx(0,1)
print(qc.draw())
state=Statevector.from_instruction(qc)
print(state)
```
