### What is a Hamiltonian?

In quantum mechanics, the Hamiltonian represents the total energy of a system. It is commonly represented by the operator $\hat{H}$ and can be formulated using Pauli operators.

### Sparse Pauli Operators

A `SparsePauliOp` is a weighted sum of Pauli strings. For example:
$$H = 0.5X + 0.3Z$$

```python
#METHOD 1
!pip install qiskit
from qiskit.quantum_info import SparsePauliOp
H=SparsePauliOp(
    ['X','Y'],
    coeffs=[0.5,0.3]
)
print(H)
```

```python
#METHOD 2
from qiskit.quantum_info import SparsePauliOp
H=SparsePauliOp.from_list([
    ('X',0.5),
    ('Y',0.3)
])
print(H)
```

### Multi-Qubit Pauli Operators

We can represent multi-qubit operators by combining single-qubit Pauli operators into a tensor product string (e.g., `'XI'`, `'ZY'`).

```python
!pip install qiskit
from qiskit.quantum_info import SparsePauliOp
H=SparsePauliOp.from_list([
    ("XI",0.1),
    ("ZY",0.3)
])
print(H.to_matrix())
```

### Comparing Qiskit Operators with Manually Constructed Matrices

Let's verify our understanding by comparing the matrix representation of a Qiskit `SparsePauliOp` against a manually constructed matrix.

```python
from qiskit.quantum_info import SparsePauliOp
import numpy as np
X=np.array([
    [0,1],
    [1,0]
])
Z=np.array([
    [1,0],
    [0,-1]
])
Ham=X*3+Z*2
print(Ham)
H=SparsePauliOp.from_list([
    ("X",3),
    ("Y",4)
])
h_matrix=H.to_matrix()
print(H)
print("Are they equal")
print(np.allclose(h_matrix,H))
```

### Inspecting the Hamiltonian Properties

You can query various properties of a `SparsePauliOp` object, such as the number of qubits, coefficients, and the underlying Pauli operators.

```python
from qiskit.quantum_info import SparsePauliOp
H=SparsePauliOp(
    ['X','Y'],
    coeffs=[0.5,0.3]
)
print(H)
print("No of qubits",H.num_qubits)
print("coefficient",H.coeffs)
print("Size of hamiltonian",H.size)
print("Pauli operators",H.paulis)
```

### Mathematical Operations

We can perform standard algebraic operations like addition directly on `SparsePauliOp` instances.

```python
from qiskit.quantum_info import SparsePauliOp
H1=SparsePauliOp.from_list([
    ("X",3),
    ("Y",4)
])
H2=SparsePauliOp.from_list([
    ("X",2),
    ("Y",3)
])
H3=H1+H2
print(H3.simplify())
```

### Scalar Multiplication

```python
from qiskit.quantum_info import SparsePauliOp
H1=SparsePauliOp.from_list([
    ("X",3),
    ("Y",4)
])
H=2*H1
print(H)
```

### Tensor Product

```python
from qiskit.quantum_info import SparsePauliOp
H1=SparsePauliOp.from_list([
    ("X",3),
    ("Y",4)
])
H2=SparsePauliOp.from_list([
    ("X",2),
    ("Y",3)
])
H=H1.tensor(H2)
print(H)
```

### Matrix Multiplication (Composition)

```python
from qiskit.quantum_info import SparsePauliOp
H1=SparsePauliOp.from_list([
    ("X",3),
    ("Y",4)
])
H2=SparsePauliOp.from_list([
    ("X",2),
    ("Y",3)
])
H=H1@H2
print(H)
```

### End-to-End Workflow

Let's combine everything into a complete flow:
1. Create a quantum circuit.
2. Apply a Hadamard gate.
3. Generate a quantum state.
4. Define a Hamiltonian.
5. Calculate its expectation value.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import SparsePauliOp,Statevector
qc=QuantumCircuit(1)
qc.h(0)
print(qc.draw())
H=SparsePauliOp.from_list([
    ("X",3),
    ("Y",0.2)
])
psi=Statevector.from_instruction(qc)
expectation=psi.expectation_value(H)
print("Expectation value",expectation)
```
